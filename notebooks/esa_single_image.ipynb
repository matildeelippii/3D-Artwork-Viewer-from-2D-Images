```python
import os 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image, ImageStat
```

```python
with Image.open("../data/raw_images/img_10.png") as img:
    img_rgb = img.convert('RGB')
    width, height = img_rgb.size
    aspect_ratio = width/height
    print(f"Larghezza: {width}, Altezza: {height}, Aspect Ratio: {aspect_ratio}")
    
    #RGB channels' means
    stats = ImageStat.Stat(img_rgb)
    mean_r, mean_g, mean_b = stats.mean
    
    #Black and white images have a single channel, so we extract the firt element of the list
    img_gray = img.convert('L')
    gray_stat = ImageStat.Stat(img_gray)
    brightness = gray_stat.mean[0]

    entropy = img_gray.entropy()

    print(mean_r, mean_g, mean_b, brightness, entropy)
    
```
